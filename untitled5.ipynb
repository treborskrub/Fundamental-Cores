```python

import math
import time
from dataclasses import dataclass
from typing import Dict, Any, Optional

# =====================================================================
# LAYER 1: GLOBAL TRNG ENGINE (The Micro-Climate Wave Generator)
# =====================================================================
class TRNGModulator:
    def __init__(self):
        self.phi = (1 + math.sqrt(5)) / 2
        self.golden_angle = (2 * math.pi) / (self.phi ** 2)
        self.theta_n = 0.517

    def advance_phase_clock(self, feedback_pulse: float = 0.0) -> float:
        """Warps the clock phase. A higher golden angle multiplier forces rapid oscillations."""
        # Multiplied by 3.5 here to simulate a highly volatile, rapidly shifting environment
        self.theta_n = (self.theta_n + (self.golden_angle * 3.5) + feedback_pulse) % (2 * math.pi)
        return self.theta_n

    def get_dynamic_target(self) -> float:
        """Calculates the breathing target invariant between 0.05 and 0.95."""
        raw_target = abs(math.sin(self.theta_n) * math.cos(self.theta_n)) * 2.0
        # Clamp to avoid total absolute zero dead-zones
        return round(max(0.05, min(0.95, raw_target)), 4)

# =====================================================================
# LAYER 3: COGNITIVELY AWARE STRUCTURAL AGENTS
# =====================================================================
@dataclass
class PatternPayload:
    content: str
    structural_density: float

class IntegratedAgent:
    def __init__(self, agent_id: str, profile: str, initial_density: float, orchestrator):
        self.agent_id = agent_id
        self.profile = profile                       # "Resonant", "Sluggish", or "Rigid"
        self.structural_density = initial_density
        self.cognitive_phase = "Liquid"
        self.handshake_failures = 0
        self.orchestrator = orchestrator

        # Profile-Specific Cognitive Adaptation Rates
        if profile == "Resonant":
            self.adaptation_rate = 0.85   # Fast, agile adaptation to clock waves
        elif profile == "Sluggish":
            self.adaptation_rate = 0.20   # Lagged, slow morphing speed
        else: # "Rigid"
            self.adaptation_rate = 0.00   # Totally blind to environmental changes

    def adapt_to_clock(self):
        """Awareness Loop: Read environmental target wave and attempt to shape-shift."""
        if self.cognitive_phase == "Gaseous":
            return # Quarantine locks out self-tuning mechanisms

        target = self.orchestrator.modulator.get_dynamic_target()

        # Ideal individual density assuming a perfectly symmetric peer distribution (a^2 + b^2 = target)
        ideal_density = math.sqrt(target / 2.0)

        # Proactive Shape-Shifting Equation with Inertia Friction
        previous_density = self.structural_density
        self.structural_density += (ideal_density - self.structural_density) * self.adaptation_rate
        self.structural_density = round(self.structural_density, 4)

        if self.adaptation_rate > 0 and abs(previous_density - self.structural_density) > 0.01:
            print(f"  -> [{self.agent_id} ({self.profile})] Morphing density: {previous_density:.3f} -> {self.structural_density:.3f} (Ideal: {ideal_density:.3f})")

    def emit_pattern(self, text: str) -> PatternPayload:
        return PatternPayload(content=text, structural_density=self.structural_density)

# =====================================================================
# LAYER 2: THE RE-ENGINEERED EVENT-DRIVEN META-AGENT
# =====================================================================
class ReactivePatternMetaAgent:
    def __init__(self):
        self.modulator = TRNGModulator()
        self.agent_registry: Dict[str, IntegratedAgent] = {}
        self.tolerance = 0.06

    def register_agent(self, agent: IntegratedAgent):
        self.agent_registry[agent.agent_id] = agent

    def broker_pattern_match_event(self, sender: IntegratedAgent, receiver: IntegratedAgent, payload: PatternPayload) -> bool:
        current_target = self.modulator.get_dynamic_target()
        print(f"\n[META-AGENT GATE] Handshake: [{sender.agent_id}] -> [{receiver.agent_id}] | Current Target Invariant: {current_target}")

        if "Gaseous" in [sender.cognitive_phase, receiver.cognitive_phase]:
            print("  -> [BLOCKED] Communication rejected. A node is unstable/Gaseous.")
            return False

        a = payload.structural_density
        margin = current_target - (a ** 2)

        if margin < 0:
            print(f"  -> [CRITICAL ASYMMETRY] Node [{sender.agent_id}] is too dense ({a}^2) for Target Wave ({current_target}).")
            self.trigger_anomaly_intervention(sender, "GEOMETRIC_CONTRADICTION")
            return False

        b = round(math.sqrt(margin), 4)
        combined_density_sq = (sender.structural_density ** 2) + (receiver.structural_density ** 2)
        deviation = abs(combined_density_sq - current_target)

        if deviation <= self.tolerance:
            print(f"  -> [MATCH SUCCESSFUL] Perfect fit: {sender.structural_density}^2 + {receiver.structural_density}^2 = {combined_density_sq:.3f}")
            sender.cognitive_phase = "Solid"
            receiver.cognitive_phase = "Solid"
            sender.structural_density = b
            receiver.structural_density = a
            sender.handshake_failures = 0
            receiver.handshake_failures = 0
            return True
        else:
            print(f"  -> [MATCH FAILED] Combined density mismatch. Deviation: {deviation:.4f}")
            receiver.handshake_failures += 1
            if receiver.handshake_failures >= 2: # Lowered to 2 for prompt ecological purging
                self.trigger_anomaly_intervention(receiver, "COMMUNICATION_LOCKOUT")
            return False

    def trigger_anomaly_intervention(self, agent: IntegratedAgent, event_type: str):
        print(f"\n[META-AGENT INTERVENTION] Purging unstable element [{agent.agent_id}] | Reason: {event_type}")
        agent.cognitive_phase = "Gaseous"

        # Bleed density as a penalty pulse to warp environment spacetime coordinates
        mutation_pulse = (agent.structural_density * 0.15) * -1.0
        new_phase = self.modulator.advance_phase_clock(feedback_pulse=mutation_pulse)

        # Cool down and stabilize to current environment baseline standard
        agent.structural_density = round(math.sqrt(self.modulator.get_dynamic_target() / 2.0), 4)
        agent.handshake_failures = 0
        agent.cognitive_phase = "Liquid"
        print(f"  -> [{agent.agent_id}] Reset to baseline density {agent.structural_density}. Restabilized to LIQUID.\n")

# =====================================================================
# RUNTIME ENVIRONMENT SIMULATION
# =====================================================================
if __name__ == "__main__":
    meta_kernel = ReactivePatternMetaAgent()

    # Instantiate the species ecosystem
    resonant = IntegratedAgent("Agent-Resonant", "Resonant", initial_density=0.4, orchestrator=meta_kernel)
    sluggish = IntegratedAgent("Agent-Sluggish", "Sluggish", initial_density=0.1, orchestrator=meta_kernel)
    rigid    = IntegratedAgent("Agent-Rigid",    "Rigid",    initial_density=0.7, orchestrator=meta_kernel)

    meta_kernel.register_agent(resonant)
    meta_kernel.register_agent(sluggish)
    meta_kernel.register_agent(rigid)

    # Begin the evolutionary time cycles
    for cycle in range(1, 4):
        print(f"\n==========================================")
        print(f" ENVIRONMENT CYCLE {cycle}")
        print(f"==========================================")

        # 1. Advance the global weather clock environment
        meta_kernel.modulator.advance_phase_clock()
        print(f"[ENVIRONMENT WATCH] Global target wave shifted to: {meta_kernel.modulator.get_dynamic_target()}")

        # 2. Awareness Phase: Agents evaluate context and mutate their shapes
        print("\n--- AWARENESS INTERFERENCE LAYER ---")
        resonant.adapt_to_clock()
        sluggish.adapt_to_clock()
        rigid.adapt_to_clock() # Will not print anything because its rate is 0.00

        # 3. Communication Phase: Test the fitness of different matches under stress
        print("\n--- TRANSACTION GATE LAYER ---")

        # Test 1: High awareness trying to sync up with high inertia
        p_resonant = resonant.emit_pattern("Sync request")
        meta_kernel.broker_pattern_match_event(resonant, sluggish, p_resonant)

        # Test 2: High awareness trying to sync with a completely blind, rigid anchor
        p_resonant_two = resonant.emit_pattern("Sync request alternative")
        meta_kernel.broker_pattern_match_event(resonant, rigid, p_resonant_two)
```
