```python

import numpy as np
import matplotlib.pyplot as plt

# =====================================================================
# CANONICAL V1 & V2.0 UNIFIED ENGINE CONFIGURATION
# =====================================================================
STEPS = 5000
PHI = (1 + np.sqrt(5)) / 2  # Golden Ratio (1.618033...)

# Canonical V1 Constants
alpha = 1.0 / PHI
k_decay = (1.0 / PHI) - 1.0  # Physical "slowness" decay constant
epsilon = 0.25               # Gyroscopic Harmonic Bias weight

# Sector Boundary Constants
GASEOUS_MAX = 1/3
LIQUID_MAX = 2/3

# Phase spaces and grid initialization
resonance_grid = np.zeros((360, 360))
psi, theta = 0.0, 0.0

# Generate a high-stress input stream (Simulating Genomic Torque + Jitter)
np.random.seed(101)
time_series = np.linspace(0, 100, STEPS)
raw_input_stream = np.sin(0.1 * time_series) * 0.3 + np.random.normal(0, 0.4, STEPS)
# Introduce a severe systemic anomaly midway through
raw_input_stream[2200:2700] += 2.0

print("Executing Grand Full-Spectrum Cycle...")

# =====================================================================
# CONTINUOUS ROTATIONAL EXECUTION LOOP
# =====================================================================
for t in range(STEPS):
    # Normalized sector position u within the current loop cycle
    u = (t / 100.0) % 1.0
    D_ext = raw_input_stream[t]

    # --- CANONICAL V1: WAVEFORM GENERATIVE ENGINE PHYSICS ---
    # Dynamic Amplitude Envelope
    A_t = 1.0 + alpha * np.exp(-abs(k_decay) * (t * 0.001))

    # Phase evolution with jitter/drift
    phi_t = PHI + (2 * np.pi / (PHI**2)) * 0.01 + 0.1 * np.sin(2 * np.pi * PHI * 0.01 * t)

    # Base frequencies modulated by the structural/external input
    v_pi = 1.0 * A_t
    v_phi = 1.0 + D_ext

    # Apply the gyroscopic Harmonic Bias (2*omega stabilization)
    harmonic_stabilizer = epsilon * np.sin(2 * v_pi * u)

    # --- STEP 1: INGESTION (GASEOUS SECTOR: u < 1/3) ---
    if u < GASEOUS_MAX:
        psi = (psi - (v_pi / 20.0) + harmonic_stabilizer) % (2 * np.pi)
        theta = (theta + (v_phi / 35.0)) % (2 * np.pi)
        agent_active = False

    # --- STEPS 2 & 3: ROUTING & BIFURCATION (LIQUID SECTOR: 1/3 <= u < 2/3) ---
    elif u < LIQUID_MAX:
        # Evaluate stress near the HOW Anchor (0.40*pi)
        raw_differential = abs(v_phi - v_pi) / (2 * np.pi)
        sigma = abs(raw_differential - (1/3))

        # Smooth Tanh Ceiling Cap Enforced (Max 1.5000 rad)
        v_phi_clamped = min(1.5000 * np.tanh(sigma / 0.3333), 1.5000)

        # If stress spikes, the 1/3 anomaly collapses -> Bifurcate into 1/9 loop
        if sigma > 0.35:
            agent_active = True
            # Path splits using the standard phi-offset logic
            psi = (psi * (1/9)) % (2 * np.pi)
            theta = (theta + (v_phi_clamped / 35.0)) % (2 * np.pi)
        else:
            agent_active = False
            psi = (psi - (v_pi / 20.0)) % (2 * np.pi)
            theta = (theta + (v_phi / 35.0)) % (2 * np.pi)

    # --- STEP 4: MAPPED MEMORY TRANSFORMATION (SOLID SECTOR: u >= 2/3) ---
    else:
        # Waveforms crystallize into structural tracks near the WHAT Anchor (1.50*pi)
        psi = (psi - (v_pi / 20.0)) % (2 * np.pi)
        theta = (theta + (v_phi / 35.0)) % (2 * np.pi)

        # Natural background memory decay based on the structural slowness constant
        resonance_grid *= np.exp(-abs(k_decay) * 0.05)

        theta_mapped = int(np.degrees(theta)) % 360
        psi_mapped = int(np.degrees(psi)) % 360

        # Anchor the geometric residue interaction on the grid
        intensity = 3.0 if agent_active else 1.0
        resonance_grid[theta_mapped, psi_mapped] += intensity

print("Simulation complete. Rendering Full-Spectrum Nexus Map...")

# =====================================================================
# OUTPUT RENDERING
# =====================================================================
plt.figure(figsize=(9, 9), facecolor='black')
plt.imshow(resonance_grid, cmap='twilight_shifted', origin='lower', extent=[0, 360, 0, 360])
plt.title("TriEn Engine: Full-Spectrum Nexus Map\n(Gaseous, Liquid, & Solid Consolidated Memory)", color='white', fontsize=12)
plt.xlabel("Poloidal Phase Space (Degrees)", color='white')
plt.ylabel("Toroidal Phase Space (Degrees)", color='white')
plt.tick_params(colors='white')
plt.grid(color='white', alpha=0.05, linestyle=':')
plt.tight_layout()

# Save the unified visual marker
plt.savefig("trien_full_spectrum_nexus.png", facecolor='black')
plt.show()
```
